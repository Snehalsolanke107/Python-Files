Types of parameter

Types of Arguments
1. postional Arguments

    2. keyword variable length Arguments

```python
# positional parameter / arguments

def Addition(num1, num2): # parameters
    add = num1 + num2
    print(add)

Addition(10,20) # arguments
Addition(50,70)
```

```python

def Addition(num1, num2): # parameters
    add = num1 + num2
    print(add)
    print(num1 , num2)

Addition(num1 = 10,num2 = 20) # arguments
Addition(num2 = 50, num1 = 70)
```

```python

def Addition(num1, num2 = 30): # parameters
    add = num1 + num2
    print(add)

Addition(10,20) # arguments
Addition(50)
```

```python

def Addition(*add): # parameters
    print(add)

Addition(10,20,30,50,40,60) # arguments
```

```python
def Std_info(**age):
    print(age)

Std_info(Roshan= 25, roshani=30, Pratik= 23)
```

```python

def Addition(num1, num2,/): # parameters
    add = num1 + num2
    print(add)

Addition(10,20) # arguments
Addition(num1 = 50, num2 = 70) # TypeError: Addition() got some positional-only arguments passed as keyword arguments: 'num1, num2'
```

```python
def Addition(*,num1, num2): # parameters
    add = num1 + num2
    print(add)

Addition(num1 =10, num2= 20) # arguments
Addition(50, 20) # TypeError: Addition() takes 0 positional arguments but 2 were given
```
