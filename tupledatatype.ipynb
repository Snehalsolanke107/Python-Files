Tuple Datatype
-- order  /sequence 
-- hectrogenious collection of elements 
-- declare inside  () parenthisis seprated by (,)
-- allow duplicate
-- immutable 
-- we can access  data using indexing and slicing
-- we can create empty tuple using ()
        eg. l1 = () # empty list
-- packing multiple variable convert into tuple
-- unpacking tuple convert into variable 

```python
t1 = ()
print(type(t1))
```

```python
t2 = (10,10.2,10+20j,"Ritik",[1,2,3],True,(1,2,3),{1,2,3},{1:1,2:4}, None)
print(t2)
```

```python
t3 = (10,20,30,10,10)
print(t3)
```

```python
print(t3.index(10, 2))
```

```python
print(t3.count(10))
```

```python
print(t3[1])
```

```python
t3[1] = 30
```

```python
l1 = [10,20,30]
print(l1)
l1[1]= 100
print(l1)
```

```python
# packing
num1 , num2 , num3 = 10,20,30

t4 = (num1 , num2 , num3)
print(t4)
print(type(t4))
```

```python
# unpacking

t5 = (100,200,300)

num1 , num2 , num3 = t5 
print(num1)
print(num2)
print(num3)
```
