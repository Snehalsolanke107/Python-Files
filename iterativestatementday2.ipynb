Question 23 — Airline Seat Reservation
An aircraft has rows and seats.
Business Context
Reservation system shows all seats.
Problem Statement
Display seat layout using nested loops.
Required Inputs
Rows
Seats per row
Expected Output
Seat identifiers.
Constraints
Nested loops only.
Learning Objective
2-dimensional iteration.
Real-world Application
Airline booking.


```python
rows = int(input("Enter rows : "))
seates = int(input("Enter a seats : "))
for row in range(1, rows + 1):
    print(f"row {row} : ", end= " ")
    
    for seat in range(1, seates + 1) :
        print(f"seat {seat}" , end= " ")
    print()

```

```python
rows = int(input("Enter rows : "))
column = int(input("Enter a column : "))
for row in range(1, rows + 1):
    for col in range(1, column + 1) :
        print("*" , end= " ")
    print()
       
```
