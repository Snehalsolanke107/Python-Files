Dict Datatype
-- we can store key and values format
-- it is map datatype
-- it is order  after version 3.7 
-- it is unorder before version 3.7
-- key is unique
-- value allow duplicate
-- we can fetch value using key
-- it is mutable
-- inside {} seprated by (,)
-- we can create empty dict using {} braces
-- key is hectrogenious collection of immutable datatype
-- value is hectrogenious collection of elements

```python
d2 = {3:9, "name":"Roshan", True: False, 2.0:4.0, 10+20j: 20+30j }

print(d2)
print(type(d2))
```

```python
d3 = {(10,20,30): (100,400,900)}
print(d3)
print(type(d3))

# TypeError: unhashable type: 'list'
# TypeError: unhashable type: 'set'
# TypeError: unhashable type: 'dict'
```
