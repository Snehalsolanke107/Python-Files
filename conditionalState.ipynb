Conditional Satement

```python
# IF Keyword

num1 = 18
if num1 >= 18 :
    print("age grater than 18")
```

```python
# If-else Conditional Statement
num1 = 18
if num1 >= 18 :
    print("age is grater than 18")
    
else :
    print("age is less than 18")
```

```python
# If-elif-else (If ladder)
age = 75
if age > 18 and age < 70 :
    print("get licence")
elif age > 70 :
    print("age is grater for licence")
else : 
    print("age less than 18")
```

```python
# Nested If Statement
marks  = -110
if marks > 0  and marks <= 100:
    if marks >= 90  and marks <= 100 :
        print("Grade A")
    elif marks >= 75 and marks < 90 :
        print("Grade B")
    elif marks >= 55 and marks < 75 :
        print("Grade C")
    else :
        print("Fail")
else :
    print("Enter valid marks (grater than 0)")
```

```python
# Match Case 
Num1 = 3 

match(Num1) :
    case 1 :
        print("one")
    case 2 :
        print("two")
    case _ :
        print("please enter valid input")
    
```
