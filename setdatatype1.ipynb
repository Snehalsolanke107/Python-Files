Set datatype
-- it is unorder dataype
-- we can store hectrogenious collection of immutable  datatype
-- enclosed by {} seprated by(,)
-- we can create empty set using set() class 
-- we can not perform indexing and slcing on set datatype
-- duplicate not allow
-- it is mutable
-- we can pass sequence value in set that time set is order

```python
s1 = set()  # empty set 
print(s1)
print(type(s1))
```

```python
s1.add(10)
print(s1)
```

```python
s1.add(10.2)
print(s1)
```
