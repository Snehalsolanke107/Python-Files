Set datatype
-- it is unorder dataype
-- we can store hectrogenious collection of immutable  datatype
-- enclosed by {} seprated by(,)
-- we can create empty set using set() class 
-- we can not perform indexing and slcing on set datatype
-- duplicate not allow
-- it is mutable
-- we can pass sequence value in set that time set is order

```python
s1 = set()  # empty set 
print(s1)
print(type(s1))
```

```python
s1.add(10)
print(s1)
```

```python
s1.add(10.2)
print(s1)
```

```python
print(s1[0]) # TypeError: 'set' object is not subscriptable
```

```python
s2 = {10,10.2,"Pratik",10+20j, True,(1,2,3), None}
print(s2)
print(type(s2))
# TypeError: unhashable type: 'list'
# TypeError: unhashable type: 'set'
# TypeError: unhashable type: 'dict'
```

```python
s3 = {10,20,30,10,10}
print(s3)     # {10,20,30}
print(type(s3))
```

```python
s3.clear()
print(s3)
```

```python
s4 = {1,2,3,4,5,6}
print(s4)
```

```python
s5 = {1,11,25,36,85,60,78}
print(s5)
```

```python
del s5
print(s5)
```

```python
s6 = {10,30,100,300,700,1000 , "A", "C", True, 10+20j}
print(s6)
```

```python
print(s6.pop())
print(s6.pop())
print(s6)
```

```python
print(s6.discard(10000))
print(s6)
```

```python
print(s6.remove(7000))
print(s6)
```
