Data add in list
-- append : last data add
-- insert  :perticular index data

```python
list5.append(10)
print(list5) # [10,20,30,10]
```

```python
list5.append(40 , 50)  # TypeError: list.append() takes exactly one argument (2 given)
print(list5) # [10,20,30,10]
```

```python
list5.append([40 , 50])
print(list5) # [10,20,30,10, [40,50]]
```
